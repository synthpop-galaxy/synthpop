```python
import synthpop 
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from astroquery.gaia import Gaia
import astropy.units as u
from astropy.coordinates import SkyCoord
from astroquery.gaia import Gaia
```

```python
synthdata = synthpop.SynthPop('huston2025_defaults.synthpop_conf',
                         chosen_bands = ['Gaia_G_EDR3', 'Gaia_BP_EDR3', 'Gaia_RP_EDR3'],
                         maglim = ['Gaia_G_EDR3', 20, "remove"],
                         post_processing_kwargs=[{"name": "ProcessDarkCompactObjects","remove": True}],
                         name_for_output='mod2test'
                        )
```

```python
synthdata.init_populations()

l_vals = np.arange(-9, -7, 1)
b_vals = np.arange(-9, -8, 1)
bin_edges = np.linspace(-30, 30, 100)

fig, axes = plt.subplots(len(b_vals), len(l_vals), figsize=(6, 6), sharex=True, sharey=True, squeeze=False)
fig.suptitle("Comparison of Gaia vs. Synthpop Proper Motions (mul / pm_l*)", fontsize=16)

Gaia.MAIN_GAIA_TABLE = "gaiadr3.gaia_source"
Gaia.ROW_LIMIT = -1

for i, b in enumerate(b_vals):
    for j, l in enumerate(l_vals):
        ax = axes[i, j]

        # GAIA 
        center = SkyCoord(l=l*u.deg, b=b*u.deg, frame='galactic').icrs
        radius = 0.1 * u.deg

        job = Gaia.cone_search_async(center, radius)
        result = job.get_results()

        mask = (result['phot_g_mean_mag'] < 20)
        filtered = result[mask]

        ra_vals = np.array(filtered['ra'], dtype=float)
        dec_vals = np.array(filtered['dec'], dtype=float)
        pmra_vals = np.array(filtered['pmra'], dtype=float)
        pmdec_vals = np.array(filtered['pmdec'], dtype=float)

        coords = SkyCoord(ra=ra_vals*u.deg,
                          dec=dec_vals*u.deg,
                          pm_ra_cosdec=pmra_vals*u.mas/u.yr,
                          pm_dec=pmdec_vals*u.mas/u.yr,
                          frame='icrs')

        gal_coords = coords.transform_to('galactic')
        pm_l_cosb = gal_coords.pm_l_cosb.to(u.mas/u.yr).value ### CHANGE HERE FOR B INSTEAD OF L

        # SYNTHPOP 
        cat2, distr2 = synthdata.process_location(
            l_deg=l, b_deg=b, solid_angle=0.01*np.pi, solid_angle_unit='deg^2'
        )
        mul_vals = cat2['mul'].astype(float).values ### CHANGE HERE FOR B INSTEAD OF L

        # PLOT
        ax.hist(pm_l_cosb, bins=bin_edges, histtype='step', color='#1f77b4', linewidth=2, label='Gaia')
        ax.hist(mul_vals, bins=bin_edges, histtype='step', color='#ff7f0e', linewidth=2, label='Synthpop')
        ax.set_title(f"l={l}, b={b}", fontsize=8)
        ax.grid(True)

        if i == len(b_vals) - 1:
            ax.set_xlabel('Proper Motion (mas/yr)', fontsize=7)
        if j == 0:
            ax.set_ylabel('N Stars', fontsize=7)
        if i == 0 and j == 0:
            ax.legend(loc='upper left', frameon=True, fontsize=6)

        ax.set_xlim(-30, 30)
        #ax.set_ylim(0, ax.get_ylim()[1])
        ax.set_box_aspect(1)
        
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
synthdata.init_populations()

l_vals = np.arange(-10, 11, 1)
b_vals = np.arange(-10, 6, 1)
#bin_edges = np.linspace(-30, 30, 100)

#fig, axes = plt.subplots(len(b_vals), len(l_vals), figsize=(6, 6), sharex=True, sharey=True, squeeze=False)
#fig.suptitle("Comparison of Gaia vs. Synthpop Proper Motions (mul / pm_l*)", fontsize=16)

Gaia.MAIN_GAIA_TABLE = "gaiadr3.gaia_source"
Gaia.ROW_LIMIT = -1

stats_list = []

for i, b in enumerate(b_vals):
    for j, l in enumerate(l_vals):
        stats_entry = {'l': l, 'b': b}
        try:
            # GAIA
            center = SkyCoord(l=l*u.deg, b=b*u.deg, frame='galactic').icrs
            radius = 0.1 * u.deg

            job = Gaia.cone_search_async(center, radius)
            result = job.get_results()

            mask = (result['phot_g_mean_mag'] < 20)
            filtered = result[mask]
            valid_pm = (~np.isnan(filtered['pmra'])) & (~np.isnan(filtered['pmdec']))
            filtered = filtered[valid_pm]

            ra_vals = np.array(filtered['ra'], dtype=float)
            dec_vals = np.array(filtered['dec'], dtype=float)
            pmra_vals = np.array(filtered['pmra'], dtype=float)
            pmdec_vals = np.array(filtered['pmdec'], dtype=float)

            coords = SkyCoord(ra=ra_vals*u.deg,
                              dec=dec_vals*u.deg,
                              pm_ra_cosdec=pmra_vals*u.mas/u.yr,
                              pm_dec=pmdec_vals*u.mas/u.yr,
                              frame='icrs')

            gal_coords = coords.transform_to('galactic')
            pm_l_cosb = gal_coords.pm_l_cosb.to(u.mas/u.yr).value

            if len(pm_l_cosb) > 0:
                stats_entry['gaia_median'] = np.median(pm_l_cosb)
                stats_entry['gaia_std'] = np.std(pm_l_cosb)
            else:
                stats_entry['gaia_median'] = np.nan
                stats_entry['gaia_std'] = np.nan

            # SYNTHPOP
            cat2, distr2 = synthdata.process_location(
                l_deg=l, b_deg=b, solid_angle=0.01*np.pi, solid_angle_unit='deg^2'
            )
            mul_vals = cat2['mul'].astype(float).values

            if len(mul_vals) > 0:
                stats_entry['synth_median'] = np.median(mul_vals)
                stats_entry['synth_std'] = np.std(mul_vals)
            else:
                stats_entry['synth_median'] = np.nan
                stats_entry['synth_std'] = np.nan

        except Exception as e:
            stats_entry['gaia_median'] = np.nan
            stats_entry['gaia_std'] = np.nan
            stats_entry['synth_median'] = np.nan
            stats_entry['synth_std'] = np.nan

        stats_list.append(stats_entry)

stats_df = pd.DataFrame(stats_list)
stats_df = stats_df.sort_values(by=['b', 'l']).reset_index(drop=True)

print(stats_df.head())

stats_df.to_csv("gaia_vs_synthpop_stats_whole_grid.csv", index=False)
```

```python
from astroquery.gaia import Gaia
from astropy.coordinates import SkyCoord
import astropy.units as u
import numpy as np
import pandas as pd

l_vals = np.arange(-10, 11, 1)
b_vals = np.arange(-10, 6, 1)

Gaia.MAIN_GAIA_TABLE = "gaiadr3.gaia_source"
Gaia.ROW_LIMIT = -1

gaia_stats_list = []
gaia_data_list = []

for i, b in enumerate(b_vals):
    for j, l in enumerate(l_vals):
        stats_entry = {'l': l, 'b': b}
        try:
            # GAIA
            center = SkyCoord(l=l*u.deg, b=b*u.deg, frame='galactic').icrs
            radius = 0.1 * u.deg

            job = Gaia.cone_search_async(center, radius)
            result = job.get_results()

            mask = (result['phot_g_mean_mag'] < 20)
            filtered = result[mask]

            valid_pm = (~np.isnan(filtered['pmra'])) & (~np.isnan(filtered['pmdec']))
            filtered = filtered[valid_pm]

            ra_vals = np.array(filtered['ra'], dtype=float)
            dec_vals = np.array(filtered['dec'], dtype=float)
            pmra_vals = np.array(filtered['pmra'], dtype=float)
            pmdec_vals = np.array(filtered['pmdec'], dtype=float)

            coords = SkyCoord(ra=ra_vals*u.deg,
                              dec=dec_vals*u.deg,
                              pm_ra_cosdec=pmra_vals*u.mas/u.yr,
                              pm_dec=pmdec_vals*u.mas/u.yr,
                              frame='icrs')

            gal_coords = coords.transform_to('galactic')
            pm_b = gal_coords.pm_b.to(u.mas/u.yr).value

            if len(pm_b) > 0:
                stats_entry['gaia_median'] = np.median(pm_b)
                stats_entry['gaia_std'] = np.std(pm_b)
            else:
                stats_entry['gaia_median'] = np.nan
                stats_entry['gaia_std'] = np.nan

            # Save all stars with l/b tagging
            tmp_df = filtered.to_pandas()
            tmp_df['pm_b'] = pm_b
            tmp_df['grid_l'] = l
            tmp_df['grid_b'] = b
            gaia_data_list.append(tmp_df)

        except Exception as e:
            stats_entry['gaia_median'] = np.nan
            stats_entry['gaia_std'] = np.nan

        gaia_stats_list.append(stats_entry)

# Summary table
medstd_df = pd.DataFrame(gaia_stats_list)

# Full Gaia stars table
fulllist_df = pd.concat(gaia_data_list, ignore_index=True)

print(medstd_df.head())

# Save outputs to CSV
medstd_df.to_csv("gaia_pm_b_stats.csv", index=False)
fulllist_df.to_csv("gaia_pm_b_allstars.csv", index=False)
```

```python
import os
import pandas as pd
import re

data_dir = "outputfiles/default"

csv_files = [f for f in os.listdir(data_dir) if f.endswith(".csv")]

all_data = []

for file in csv_files:
    match = re.search(r"l([-+]?\d+\.\d+)_b([-+]?\d+\.\d+)", file)
    if match:
        l_val = float(match.group(1))
        b_val = float(match.group(2))
    else:
        l_val, b_val = None, None

    df = pd.read_csv(os.path.join(data_dir, file))

    if "mub" in df.columns:
        df_out = df[["mub"]].copy()
        df_out["l"] = l_val
        df_out["b"] = b_val
        all_data.append(df_out)

b_synth_combined_df = pd.concat(all_data, ignore_index=True)

b_synth_stats_df = b_synth_combined_df.groupby(["l", "b"]).agg(
    mub_median=("mub", "median"),
    mub_std=("mub", "std")
).reset_index()

# Save both files
b_synth_combined_df.to_csv("synthpop_mub_all.csv", index=False)
b_synth_stats_df.to_csv("synthpop_mub_stats.csv", index=False)

print(f"Saved combined mub data ({combined_df.shape[0]} rows) to synthpop_mub_all.csv")
print(f"Saved mub statistics ({stats_df.shape[0]} grid cells) to synthpop_mub_stats.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("l_gaia_vs_synthpop_stats_whole_grid.csv")

b_values_to_plot = [-9, -4, 0, 4]

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey=True)
axes = axes.flatten()

for idx, b_val in enumerate(b_values_to_plot):
    ax = axes[idx]
    
    df_b = df[df['b'] == b_val].sort_values(by='l')
    
    ax.errorbar(df_b['l'], df_b['gaia_median'], 
                yerr=df_b['gaia_std'], fmt='-o', label='Gaia',
                capsize=3, color='tab:blue')
    
    ax.errorbar(df_b['l'], df_b['synth_median'], 
                yerr=df_b['synth_std'], fmt='-o', label='Synthpop',
                capsize=3, color='tab:orange')
    
    ax.set_title(f"b = {b_val}°")
    ax.set_xlabel("Galactic Longitude l (deg)")
    ax.set_ylabel("Median Proper Motion (mas/yr)")
    ax.grid(True)
    if idx == 0:
        ax.legend()

fig.suptitle("Gaia vs Synthpop Median Proper Motion in l With Constant b", fontsize=16)
plt.tight_layout()
plt.show()
```
